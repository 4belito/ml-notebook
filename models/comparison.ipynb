| Algorithm | Main Update Equation | Key Idea |
|---|---|---|
| Q-learning | $Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$ | Off-policy TD learning |
| SARSA | $Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma Q(s',a') - Q(s,a)\right]$ | On-policy TD learning |
| DQN | $\theta \leftarrow \theta - \eta \nabla_\theta \left(r + \gamma \max_{a'} Q_{\theta^-}(s',a') - Q_\theta(s,a)\right)^2$ | Q-learning with neural network |
| REINFORCE | $\nabla_\theta J = E[\nabla_\theta \log \pi_\theta(a\|s) \, G_t]$ | Monte Carlo policy gradient |
| Actor-Critic | **Actor:** $\theta \leftarrow \theta + \alpha \nabla_\theta \log \pi_\theta(a\|s)\,\delta_t$ <br> **Critic:** $V(s) \leftarrow V(s) + \beta \delta_t$ <br> where $\delta_t = r + \gamma V(s') - V(s)$ | Policy gradient with TD baseline |
| PPO | update actor update $L^{PPO} = E[\min(r_t A_t,\ \text{clip}(r_t,1-\epsilon,1+\epsilon)A_t)]$ | Clipped policy gradient |
| DDPG | **Actor:** $\nabla_\theta J \approx E[\nabla_a Q(s,a)\nabla_\theta \mu_\theta(s)]$ <br> **Critic:** $L = (r + \gamma Q(s',\mu(s')) - Q(s,a))^2$ | Deterministic actor–critic |
| TD3 | **Actor:** same as DDPG <br> **Critic:** twin critics minimize $L = (r + \gamma \min(Q_1,Q_2) - Q (s,a))^2$ | Reduce Q overestimation |
| SAC | **Actor:** $J_\pi = E[\alpha \log \pi(a\|s) - Q(s,a)]$ <br> **Critic:** $L = (r + \gamma V(s') - Q(s,a))^2$ | Maximum entropy RL |

| Algorithm | Objective Function | Key Idea |
|---|---|---|
| Q-learning | $J(Q) = \mathbb{E}\left[(r + \gamma \max_{a'} Q(s',a') - Q(s,a))^2\right]$ | Off-policy TD learning |
| SARSA | $J(Q) = \mathbb{E}\left[(r + \gamma Q(s',a') - Q(s,a))^2\right]$ | On-policy TD learning |
| DQN | $J(\theta) = \mathbb{E}\left[(r + \gamma \max_{a'} Q_{\theta^-}(s',a') - Q_\theta(s,a))^2\right]$ | Q-learning with neural network |
| REINFORCE | $J(\theta) = \mathbb{E}\left[\log \pi_\theta(a\|s)\, G_t\right]$ | Monte Carlo policy gradient |
| Actor-Critic | **Actor:** $J_\pi(\theta) = \mathbb{E}\left[\log \pi_\theta(a\|s)\, A_t\right]$ <br> **Critic:** $J_V(\phi) = \mathbb{E}\left[(r + \gamma V_\phi(s') - V_\phi(s))^2\right]$ | Policy gradient with TD baseline |
| PPO | **Actor:** $J_\pi(\theta) = \mathbb{E}\left[\min(r_t A_t,\ \text{clip}(r_t,1-\epsilon,1+\epsilon)A_t)\right]$ <br> **Critic:** $J_V(\phi) = \mathbb{E}\left[(r + \gamma V_\phi(s') - V_\phi(s))^2\right]$ | Clipped policy gradient |
| DDPG | **Actor:** $J_\pi(\theta) = \mathbb{E}[Q_\phi(s,\mu_\theta(s))]$ <br> **Critic:** $J_Q(\phi) = \mathbb{E}\left[(r + \gamma Q_{\phi^-}(s',\mu_{\theta^-}(s')) - Q_\phi(s,a))^2\right]$ | Deterministic actor–critic |
| TD3 | **Actor:** $J_\pi(\theta) = \mathbb{E}[Q_{\phi_1}(s,\mu_\theta(s))]$ <br> **Critic:** $J_Q(\phi_i) = \mathbb{E}\left[(r + \gamma \min(Q_{\phi_1^-},Q_{\phi_2^-}) - Q_{\phi_i}(s,a))^2\right]$ | Reduce Q overestimation |
| SAC | **Actor:** $J_\pi(\theta) = \mathbb{E}[\alpha \log \pi_\theta(a\|s) - Q_\phi(s,a)]$ <br> **Critic:** $J_Q(\phi) = \mathbb{E}\left[(r + \gamma V(s') - Q_\phi(s,a))^2\right]$ | Maximum entropy RL |

where $A(s,a) = Q(s,a) - V(s)$

## Model-based RL algorithms using Bellman equations

- **Value Iteration**
  - Repeatedly applies the Bellman optimality update:
  - $V_{k+1}(s) = \max_a \left[R(s,a) + \gamma \sum_{s'} P(s' \mid s,a) V_k(s')\right]$

- **Policy Iteration**
  - Alternates two steps:
    - **Policy evaluation**
      - $V^\pi(s) = R(s,\pi(s)) + \gamma \sum_{s'} P(s' \mid s,\pi(s)) V^\pi(s')$
    - **Policy improvement**
      - $\pi'(s) = \arg\max_a \left[R(s,a) + \gamma \sum_{s'} P(s' \mid s,a) V^\pi(s')\right]$

- **Modified Policy Iteration**
  - Hybrid between value iteration and policy iteration.
  - Performs partial policy evaluation before policy improvement.

- **Dyna-Q**
  - Learns a model $\hat P(s' \mid s,a)$ and $\hat R(s,a)$.
  - Uses simulated transitions to perform additional Bellman updates.

- **Model Predictive Control (MPC)**
  - Uses a learned dynamics model
  - $s_{t+1} = f_\theta(s_t,a_t)$
  - Optimizes a sequence of actions through planning.

- **MuZero / Dreamer (modern model-based RL)**
  - Learn latent dynamics models
  - Perform planning using Monte Carlo Tree Search or imagined rollouts.
