## Tokenizer

The tokenizer is a `WordLevel` model trained on the translation corpus. It maps words to integer IDs. Unknown or rare words (below `min_frequency=2`) map to `[UNK]`.

```python
from tokenizers import Tokenizer

tok = Tokenizer.from_file("tokenizer_en.json")
```

### Vocabulary

```python
print("Vocab size:", tok.get_vocab_size())

# Special tokens and their IDs
vocab = tok.get_vocab()
special = {k: v for k, v in vocab.items() if k.startswith("[") and k.endswith("]")}
print("Special tokens:", special)

# Lookup by token or ID
print(tok.token_to_id("[SOS]"))  # → 2
print(tok.id_to_token(2))  # → "[SOS]"
print(tok.token_to_id("of"))  # → 10 (example)
print(tok.id_to_token(10))  # → "hello" (example)
```

### Encode & Decode

`encode` splits a sentence into tokens and returns their IDs. Words not in the vocabulary are replaced by `[UNK]`.

```python
enc = tok.encode("Hello, how are you?")
print("Tokens:", enc.tokens)  # word pieces
print("IDs:   ", enc.ids)  # integer IDs
print("Decoded:", tok.decode(enc.ids))  # back to string
```

### Batch Encode & Decode

```python
encs = tok.encode_batch(["How are you?", "Good morning"])
for enc in encs:
    print(enc.tokens, "→", enc.ids)

tok.decode_batch([e.ids for e in encs])
```
